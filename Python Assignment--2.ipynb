```python
age_list = [24, 25, 26, 27, 28]
name_list = ["Anu", "Binu", "Cera", "Diya", "Elza"]
print(age_list)
print(name_list)
```

```python
# a. Append "Yazhini" to name_list
name_list.append("Yazhini")
print("After append:", name_list)

# b. Insert 30 at index 2 in age_list
age_list.insert(2, 30)
print("After insert:", age_list)

# c. Remove "Yazhini" from name_list
name_list.remove("Yazhini")
print("After remove:", name_list)

# d. Pop the last element from age_list
age_list.pop()
print("After pop:", age_list)

# e. Extend age_list with [29, 30, 26]
age_list.extend([29, 30, 26])
print("After extend:", age_list)

# f. Sort age_list in descending order
age_list.sort(reverse=True)
print("After sort (descending):", age_list)

# g. Find Max, Min and Sum of age_list
print("Max age:", max(age_list))
print("Min age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

```python
# 3. Accessing List Elements
print("First element of name_list:", name_list[0])
print("Last element of name_list:", name_list[-1])
print("Elements from index 2 to 4 in name_list:", name_list[2:5])
print("Elements of name_list in reverse order:", name_list[::-1])

# Dictionary (Creation, Modification and Access)
student_marks = {"Anu": 85, "Binu": 90, "Cera": 78, "Diya": 92, "Elza": 88}
print("Student marks dictionary:", student_marks)

print("Mark of Anu:", student_marks["Anu"])

student_marks["Janani"] = 80
print("After adding Janani:", student_marks)

student_marks["Anu"] = 82
print("After updating Anu's mark:", student_marks)

print("Keys:", student_marks.keys())
print("Values:", student_marks.values())
print("Items:", student_marks.items())
```
